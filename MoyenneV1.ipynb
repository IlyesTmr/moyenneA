```python
import numpy as np
import matplotlib.pyplot as plt

matiere = [f"R{i}" for i in range (101,116)]+[f"SAE{i}" for i in range(11,17)] #definition de matière
note=[i for i in range(21)]
coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

diconote={cle : valeur for cle,valeur in zip(matiere,note)}
dicoef1={cle : valeur for cle,valeur in zip(matiere,coef1)}
dicoef2={cle : valeur for cle,valeur in zip(matiere,coef2)}
dicoef3={cle : valeur for cle,valeur in zip(matiere,coef3)}


somme=0
somme_coef=0

for elt in diconote.keys():
    somme+=diconote[elt]*dicoef1[elt]
    somme_coef+=dicoef1[elt]
    moy1=(somme/somme_coef)
print("La moyenne en UE1 est :" ,moy1)


for elt in diconote.keys():
    somme+=diconote[elt]*dicoef2[elt]
    somme_coef+=dicoef2[elt]
    moy2=(somme/somme_coef)
print("La moyenne en UE2 est :" ,moy2)


for elt in diconote.keys():
    somme+=diconote[elt]*dicoef3[elt]
    somme_coef+=dicoef3[elt]
    moy3=(somme/somme_coef)
print("La moyenne en UE3 est :" ,moy3)

def code_couleur(val):
    if val>10:
        return"green"
    elif 8<=val<10:
        return"orange"
    elif val<8:
        return"red"
colors=[code_couleur(val) for val in [moy1,moy2,moy3]]
plt.bar(["UE1","UE2","UE3"],[moy1,moy2,moy3],color=colors)
```
