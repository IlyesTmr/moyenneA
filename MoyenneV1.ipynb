Calcul Moyenne UE

```python
somme=0
somme_coef=0

for i in range (101, 104):
    note = float(input("Indiquez votre note dans la ressource R{}".format(i)))
    coefficient = float(input("Indiquez le coefficient dans la ressource R{}".format(i)))
    somme+=note*coefficient
    somme_coef=+coefficient
moy_UE1 = somme/somme_coef
print("la moyenne de l'UE1 vaut :" ,moy_UE1)

 
```
